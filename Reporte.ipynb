## Resumen
### Se realizó un recorrido por los datos de un dataset, en donde fue necesario hacer una limpieza de datos para tener mejores resultados al momento de hacer análisis o realizar graficas de los datos

## 1. Introducción
### En esta actividad se utilizó como conjunto de datos un archivo de Excel sobre compradores de bicis, sin embargo, es necesario realizar una limpieza de datos dado que hay inconsistencias que pueden afectar al momento de analizar los datos

## 1.1. Conjunto de Datos
### El conjunto de datos fue obtenido de los data sets de Kaggle, cuenta con información de clientes, tales como su edad, sus ingresos, educación, entre otros, y cuenta con más de 1000 datos. El archivo pesa 110Kb

## 1.2. Objetivos de la Limpieza
### Dado que los valores de algunos datos están alterando el resultado de las medidas de tendencia del conjunto de datos.

## 2. Exploración Inicial
### Al principio se encontró que los valores que se encontraban en la columna de “Income” al contar con una coma, no lo tomaba como un valor numérico, sino como un string, por lo que habría que cambiarlo.
### Se encontraron filas que se repetían, pues se encontraban exactamente los mismos datos que en otras.
### Se encontraron valores nulos entre diferentes filas y columnas.

## 3. Proceso de Limpieza
### Se estandarizó la forma de colocar los valores en la columna “Income”, removiendo el “,” del valor.
### Se eliminaron las filas que estaban duplicadas usando drop_duplicates()
### Se eliminaron las filas con valores nulos usando dropna()
### Se creó un nuevo archivo con los valores que aun se conservaban del dataset

## 4. Resultados
### Nos quedamos con 952 filas con sus valores completos y sin repetir filas, Las medidas de tendencia cambiaron un poco con respecto al conjunto de datos antes de la limpieza.
## 5. Conclusiones
### Mantener un conjunto de datos limpio ayuda a que sea más confiable y sus datos más representativos, así como también ayudan a la precisión de los análisis.

## 6. Recomendaciones
### Revisar los datos si es que se actualizara la información del dataset. De igual forma es posible rellenar los datos a partir de las estadísticas de los demás datos.
