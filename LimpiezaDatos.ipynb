```python
import pandas as pd
```

### Lee el archivo e imprime información de sus datos numericos

```python
df = pd.read_csv('bike_buyers.csv')

df["Income"] = pd.to_numeric(df["Income"].str.replace(',', ''))
```

```python
print('Describe: \n', df.describe())
```

```python
df.info()
```

### ¿Datos duplicados?

```python
#Filas duplicadas
duplicados = df[df.duplicated()]

#Mostrar duplicados
print('Datos duplicados:')
print(duplicados)
```

### ¿Datos nulos?

```python
#Filas con valores nulos
nulos = df[df.isnull().any(axis=1)]

print('Datos con valores nulos:')
print(nulos)
```

### Limpieza de datos

```python
#Eliminar duplicados y nulos
df = df.drop_duplicates()
df = df.dropna()
```

### Archivo limpio

```python
#Guardar datos limpios
df.to_csv('bike_buyers_clean.csv', index=False)
print('Datos limpios guardados en bike_buyers_clean.csv')
```

```python
df_clean = pd.read_csv('bike_buyers_clean.csv')

print('Describe: \n', df_clean.describe())
```

### ¿Datos duplicados?

```python
#Filas duplicadas
duplicados = df_clean[df_clean.duplicated()]

#Mostrar duplicados
print('Datos duplicados:')
print(duplicados)
```

### ¿Datos nulos?

```python
#Filas con valores nulos
nulos = df_clean[df_clean.isnull().any(axis=1)]

print('Datos con valores nulos:')
print(nulos)
```
